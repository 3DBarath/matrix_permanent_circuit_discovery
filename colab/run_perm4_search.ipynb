# Perm_4 17-Multiplication Circuit Search (Tesla T4)

This notebook detects the Colab GPU, compiles `perm4_17mul_search.cu` with `nvcc -O3 -arch=sm_75`, and runs the CUDA search for 17-multiplication circuits computing the 4x4 matrix permanent (`Perm_4`).

```python
!nvidia-smi
!nvcc --version
```

```python
import os

candidate_paths = [
    "colab/perm4_17mul_search.cu",
    "perm4_17mul_search.cu",
]

cu_path = next((p for p in candidate_paths if os.path.exists(p)), None)

if cu_path is None:
    print("perm4_17mul_search.cu not found locally. Please upload perm4_17mul_search.cu:")
    from google.colab import files
    uploaded = files.upload()
    if "perm4_17mul_search.cu" in uploaded:
        cu_path = "perm4_17mul_search.cu"
    elif uploaded:
        cu_path = next(iter(uploaded.keys()))
    else:
        raise FileNotFoundError("No CUDA source file uploaded.")

os.environ["PERM4_CU_PATH"] = cu_path
print(f"Using CUDA source: {cu_path}")
```

```python
!nvcc -O3 -arch=sm_75 "$PERM4_CU_PATH" -o perm4_17mul_search
```
