# Gen 9 CSE+DCE Gate & Depth Verification for $n = 14$ and $n = 16$

Computes the exact `compile_opttree_slp(n)` non-scalar multiplications, CSE+DCE total arithmetic gates, and circuit depth for $n = 14$ and $n = 16$ (for Table 1 of Paper I), without stressing a local laptop.

```python
import sys, time
sys.path.insert(0, '.')
from src.slp_compiler import compile_opttree_slp

for n in [12, 14, 16]:
    t0 = time.time()
    slp = compile_opttree_slp(n)
    dt = time.time() - t0
    N = 1 << (n - 1)
    glynn_gates = n*(n-1) + n*(N-1) + (n-1)*N + (N-1) + 1
    savings = 100.0 * (slp['total_gates'] - glynn_gates) / glynn_gates
    print(f'n={n:2d} | muls={slp["non_scalar_muls"]:,} | Gen9 CSE gates={slp["total_gates"]:,} | Glynn+Gray={glynn_gates:,} ({savings:+.1f}%) | depth={slp["depth"]} | {dt:.2f}s')
```
